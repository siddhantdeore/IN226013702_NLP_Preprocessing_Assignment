1. What is the difference between "Love" and "love" in NLP?

In Natural Language Processing (NLP), "Love" and "love" are treated as two different tokens because NLP models are case-sensitive by default. This means that uppercase and lowercase versions of the same word are considered separate entries in the vocabulary. For example, "Love" might appear at the beginning of a sentence, while "love" appears elsewhere, but both will be processed differently unless text normalization techniques like lowercasing are applied. Converting all text to lowercase helps reduce redundancy and ensures that the model treats both forms as the same word, improving consistency and efficiency.

2. What happens if stopwords are not removed?

If stopwords are not removed, the dataset may contain a large number of common words such as "is", "the", "and", and "in", which do not carry significant meaning. This increases the size of the feature space and introduces noise into the data, making it harder for machine learning models to focus on important keywords. As a result, the model may become less efficient and slower to train. However, in some cases, keeping stopwords can be useful, especially when the context or sentence structure is important for understanding meaning.

3. Provide two real-world scenarios where removing stopwords can be harmful.

Removing stopwords can be harmful in certain real-world applications where these words carry important meaning. One example is sentiment analysis, where words like "not" play a crucial role in determining the sentiment of a sentence. For instance, removing "not" from "I am not happy" changes the meaning completely. Another example is in question-answering systems or chatbots, where stopwords such as "what", "when", and "how" are essential for understanding the user’s query. Removing these words can lead to incorrect interpretation and poor responses.

4. Explain the difference between stemming and lemmatization.

Stemming and lemmatization are both techniques used to reduce words to their base or root form, but they differ in their approach and accuracy. Stemming is a simple and fast method that removes prefixes or suffixes from words using basic rules, often resulting in incomplete or incorrect words, such as converting "running" to "runn". On the other hand, lemmatization is a more advanced technique that uses vocabulary and grammatical analysis to convert words into their meaningful base form, known as a lemma. For example, "running" is converted to "run" using lemmatization. While stemming is faster, lemmatization provides more accurate and meaningful results.

```python
import re

def preprocess_text(text):
    """
    Cleans and preprocesses input text.
    Returns tokens and cleaned sentence.
    """

    # -------- Error Handling --------
    if not text or not isinstance(text, str):
        return [], ""

    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r"http\S+|www\S+", "", text)

    # Remove emails
    text = re.sub(r"\S+@\S+", "", text)

    # Remove numbers
    text = re.sub(r"\d+", "", text)

    # Remove emojis and special characters
    text = re.sub(r"[^\w\s]", "", text)

    # Handle repeated characters (e.g., soooo → so)
    text = re.sub(r"(.)\1{2,}", r"\1", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    # Tokenization
    tokens = text.split()

    # Remove short tokens (<=2), but keep "no", "not"
    cleaned_tokens = [
        word for word in tokens
        if len(word) > 2 or word in ["no", "not"]
    ]

    cleaned_sentence = " ".join(cleaned_tokens)

    return cleaned_tokens, cleaned_sentence
```

```python
test_sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

for text in test_sentences:
    tokens, sentence = preprocess_text(text)

    print("\nOriginal:", text)
    print("Tokens:", tokens)
    print("Cleaned Sentence:", sentence)
```

```python
def token_analytics(tokens):
    if not tokens:
        return 0, 0, 0

    total = len(tokens)
    unique = len(set(tokens))
    avg_len = sum(len(t) for t in tokens) / total

    return total, unique, round(avg_len, 2)


for text in test_sentences:
    tokens, _ = preprocess_text(text)
    total, unique, avg = token_analytics(tokens)

    print(f"\nText: {text}")
    print(f"Total Tokens: {total}")
    print(f"Unique Tokens: {unique}")
    print(f"Avg Token Length: {avg}")
```

```python
from collections import Counter

all_tokens = []

for text in test_sentences:
    tokens, _ = preprocess_text(text)
    all_tokens.extend(tokens)

counter = Counter(all_tokens)

print("\nTop 10 Frequent Words:")
print(counter.most_common(10))

print("\nTop 5 Least Frequent Words:")
print(counter.most_common()[-5:])
```

```python
def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []

    for text in text_list:
        tokens, sentence = preprocess_text(text)

        all_tokens.extend(tokens)
        clean_sentences.append(sentence)

    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }
```

```python
edge_cases = [
    "",
    "😍😍😍",
    "1234567890"
]

for case in edge_cases:
    tokens, sentence = preprocess_text(case)
    print("\nInput:", case)
    print("Tokens:", tokens)
    print("Sentence:", sentence)
```
